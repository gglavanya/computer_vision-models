```python
import cv2
import numpy as np

cap = cv2.VideoCapture(0)
cap.set(cv2.CAP_PROP_FRAME_WIDTH, 640)
cap.set(cv2.CAP_PROP_FRAME_HEIGHT, 480)

while True:
    _, frame = cap.read()
    hsv_frame = cv2.cvtColor(frame, cv2.COLOR_BGR2HSV)
    height, width, _ = frame.shape
    cx = int(width / 2)
    cy = int(height / 2)
    pixel_center = hsv_frame[cy, cx]
    hue, sat, val = pixel_center

    # Determine color
    if val < 50:
        color = "Black"
    elif hue < 5:
        color = "Red"
    elif 5 <= hue < 20:
        color = "Orange"
    elif 20 <= hue < 30:
        color = "Yellow"
    elif 30 <= hue < 85:
        color = "Green"
    elif 85 <= hue < 125:
        color = "Blue"
    elif 125 <= hue < 150:
        color = "Purple"
    elif 150 <= hue < 170:
        color = "Pink/Peach"
    else:
        color = "Red"

    cv2.putText(frame, color, (10, 50), cv2.FONT_HERSHEY_SIMPLEX, 1, (0, 0, 0), 2)
    cv2.circle(frame, (cx, cy), 5, (255, 0, 0), 3)
    cv2.imshow("Image", frame)
    
    key = cv2.waitKey(1)
    if key == 27:  # ESC to exit
        break

cap.release()
cv2.destroyAllWindows()
cv2.waitKey(1)
```
